**Question 5:** Write a Python program to calculate the factorial of a number.

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n-1)    
    

a = int(input("enter a number to find the factorial"))
factorial(a)    




```

```python
def factorial(n):
    result = 1 
    for i in range (2,n+1):
        result *= i
    return result     

a = int(input("enter a number to find the factorial"))
factorial(a)    
```

```python
import math
a = int(input("enter a number to find the factorial"))
print(math.factorial(a)) 
```
